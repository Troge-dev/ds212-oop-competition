## 1. Encapsulation — "Protect the Object's State"

**Guiding Question:** *Who is allowed to read or change this object's internal data?*

**Key DS212 Python Tools:**
1. **Protected attributes (`_name`)**: Signal that internal state should not be mutated directly from outside.
2. **Read-only `@property`**: Expose values safely without letting callers overwrite them.
3. **Validating `@attr.setter`**: Reject invalid states immediately by raising `ValueError`.
4. **Defensive copies (`tuple`)**: Return `tuple(self._items)` instead of exposing the internal `list` directly.

```python
class SensorLog:
    def __init__(self, sensor_id: str, limit: float):
        if limit <= 0:
            raise ValueError("limit must be positive")
        self._sensor_id = sensor_id
        self._limit = limit
        self._readings = []

    @property
    def sensor_id(self) -> str:
        return self._sensor_id  # Read-only: no setter defined

    @property
    def limit(self) -> float:
        return self._limit

    @limit.setter
    def limit(self, value: float):
        if value <= 0:
            raise ValueError("limit must be positive")
        self._limit = value

    @property
    def readings(self) -> tuple:
        return tuple(self._readings)  # Tuple prevents external .append()

    def record(self, value: float):
        if value < 0 or value > self._limit:
            raise ValueError(f"Reading {value} is out of valid range [0, {self._limit}]")
        self._readings.append(value)


log = SensorLog("TEMP-01", 100.0)
log.record(28.5)
print("Readings:", log.readings)

try:
    log.limit = -10
except ValueError as err:
    print("Blocked invalid limit:", err)
```

## 2. Inheritance — "Reuse and Specialize"

**Guiding Question:** *Is this class a specialized kind of another class (`is-a` relationship)?*

**Key DS212 Python Tools:**
1. **`class Child(Parent):`**: Inherits attributes and methods from `Parent`.
2. **`super().__init__(...)`**: Calls the parent constructor so shared state is initialized once without copy-pasting.
3. **Overriding**: Redefining a method in `Child` to replace parent behavior.
4. **Extending (`super().method()`)**: Calling the parent method inside the child method and adding extra information.

```python
class StaffMember:
    def __init__(self, name: str, base_pay: float):
        self._name = name
        self._base_pay = base_pay

    def compute_pay(self) -> float:
        return self._base_pay

    def describe(self) -> str:
        return f"Staff: {self._name}"


class LabInstructor(StaffMember):
    def __init__(self, name: str, base_pay: float, lab_bonus: float):
        super().__init__(name, base_pay)  # Reuse parent setup
        self.lab_bonus = lab_bonus

    def compute_pay(self) -> float:
        # Override completely
        return self._base_pay + self.lab_bonus

    def describe(self) -> str:
        # Extend parent description using super()
        return f"{super().describe()} | Role: Lab Instructor"


prof = LabInstructor("Alex", 20000.0, 3500.0)
print(prof.describe())
print("Total pay:", prof.compute_pay())
print("Is StaffMember?", isinstance(prof, StaffMember))
```

## 3. Polymorphism — "One Interface, Many Forms"

**Guiding Question:** *Can different objects be used through the exact same method call without checking their type?*

**Key DS212 Python Tools:**
1. **Eliminate `if isinstance(...)` chains**: A loop should simply call `obj.method()` and let each object handle its own behavior.
2. **Duck Typing**: An unrelated class (that does **not** inherit from the parent) still works in the exact same loop as long as it implements the expected method name and signature.

```python
class EmailNotifier:
    def send(self, msg: str) -> str:
        return f"[EMAIL] {msg}"


class SMSNotifier:
    def send(self, msg: str) -> str:
        return f"[SMS] {msg}"


# Duck-typed class: does NOT inherit from anything above, just implements .send()
class CampusSiren:
    def send(self, msg: str) -> str:
        return f"[SIREN ALERT] {msg.upper()}"


def broadcast(channels: list, message: str):
    # Clean polymorphism: NO isinstance() checks needed
    for ch in channels:
        print(ch.send(message))


broadcast([EmailNotifier(), SMSNotifier(), CampusSiren()], "Lab starts now")
```
